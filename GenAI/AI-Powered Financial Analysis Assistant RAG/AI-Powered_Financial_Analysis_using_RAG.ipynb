# 6. Create a Financial knowledge Base


```python
knowledge = []

# Company information
knowledge.append("""
Company Name : Reliance Industries Limited
Stock Ticker : RELIANCE.NS
Data Source : Yahoo Finance
Analysis : Fundamental financial analysis
Working Sector :  Energy
Industry Name :  Oil & Gas Refining & Marketing
Website Name :  https://www.ril.com
Market Cap :  15788334383104 INR
""")

# Revenue and net income
for year in revenue.index.intersection(net_income.index) :
    rev = revenue.loc[year]
    profit = net_income.loc[year]

    if pd.notna(rev) and pd.notna(profit):
        knowledge.append(
            f"In financial year {pd.Timestamp(year).year}, "
            f"Reliance Industries reported revenue of : "
            f"Rs {rev / 10000000:.2f} crore and "
            f"net income of Rs. {profit / 10000000:.2f} crore."
        )
    

# Fonancial Rations :
for year in ratios.index:
    knowledge.append(
        f"In financial year {year}, Reliance Industries Limited had "
        f"a debt-to-equity ratio of "
        f"{ratios.loc[year, 'Debt-to-Equity']:.2f}, "
        f"a current ratio of "
        f"{ratios.loc[year, 'Current Ratio']:.2f}, "
        f"and a net profit margin of "
        f"{ratios.loc[year, 'Profit Margin (%)']:.2f} %."
    )


# Cash flow
for year in cashflow.columns:
    operating = cashflow.loc["Operating Cash Flow", year]
    free = cashflow.loc["Free Cash Flow", year]

    if pd.notna(operating) and pd.notna(free):
        knowledge.append(
            f"In financial year {pd.Timestamp(year).year}, "
            f"Reliance Industries Limited had operating cash flow of "
            f"Rs. {operating / 10000000:.2f} crore and "
            f"free cash flow of Rs. {free / 10000000:.2f} crore."
        )

    
print("Total knowledge entries : ",len(knowledge))
print("First entry : \n",knowledge[0])
```

```python
print("All Entries in knowledge base :")
for i, item in enumerate(knowledge, start = 1) :
    print(f"\n Entry no : {i}")
    print(item)
```

```python
with open("reliance_financial_knowledge.txt", "w", encoding="utf-8") as file :
    for item in knowledge :
        file.write(item + "\n\n")

print("Knowledge base saved successfully")
```

# 7. Embeddings 

```python
# Load libraries and model
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded sucessfully")
```

```python
# Generate embeddings
embeddings = model.encode(knowledge, show_process_bar = True)

print("Total Knowledge Entries : ", len(knowledge))
print("Embedding Shape : ",embeddings.shape)
```

```python
print(embeddings[0][:10])
```

# 8. Create Faiss Index

```python
import faiss 

embeddings = np.array(embeddings).astype("float32")

index = faiss.IndexFlatL2(embeddings.shape[1])

index.add(embeddings)

print("Faiss index created sucessfully")
print("Total vectore sorted : ",index.ntotal)
```

# 9. Fianancial information using question

```python
question = "How profitable is Reliance Industries?"

question_embedding = model.encode([question])

question_embedding = np.array(question_embedding).astype("float32")

print("Question embedding shape : ",question_embedding.shape)
```

```python
# Retrive most relevant 3 entries

distances, indices = index.search(question_embedding, 3)

for i in indices[0]:
    print(knowledge[i])
    print("-"*50)
```

```python
# 10. Prepare and retrived context
context = "\n".join(knowledge[i] for i in indices[0])

print(context)
```

```python
from getpass import getpass
from google import genai

api_key = getpass("Enter your Gemini API Key : ")
client = genai.Client(api_key = api_key)
```

# 11. Generate Answer using RAG

```python
question = "How did Reliance Industries' revenue and net income change from FY 2023 to FY 2025?"

prompt = f"""
You are a financial analysis assistant.

Answer the question using only the context provided.
Do not invent figures or information.

context : {context}
Question = {question}
"""

response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents = prompt
)

print(response.text)
```

# 12. Create function to ask question

```python
def ask_question(question):
    # Convert question into embedding
    question_embedding = model.encode([question]).astype("float32")

    # Retrive the information
    distances, indices = index.search(question_embedding, 3)
    context = "\n".join(knowledge[i] for i in indices[0])

    # Ask Gemini
    prompt = f"""
    Answer using only the context below.
    If information is missing, say so.

    context : {context}
    Question = {question}
    """
    response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents = prompt
    )

    print(response.text)
```

```python
# Now ask question
ask_question("What was Reliance's revenue in FY 2025?")
```

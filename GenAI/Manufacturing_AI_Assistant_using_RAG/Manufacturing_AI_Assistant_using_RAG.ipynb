```python
import os

documents = {}

for filename in os.listdir(doc_path):
    file_path = os.path.join(doc_path,filename)

    with open(file_path, "r", encoding = "UTF-8") as file :
        documents[filename] = file.read()

print("Total documents : ",len(documents))
```

```python
for filename, content in documents.items():
    print(filename, ":" ,len(content), "characters") 
```

# 3. Spliting data in to chunks

```python
chunks = []

for filename, content in documents.items():
    for i in range(0, len(content), 500):
        chunks.append({
            "source": filename,
            "text": content[i:i+500]
        })


print("Total Chunks : ",len(chunks))
```

```python
print("Source :", chunks[0]["source"])
print(chunks[0]["text"])
```

# 4. Generate embeddings

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
texts = [chunk["text"] for chunk in chunks]

embeddings = model.encode(texts)

print("Embeddings Shape : ",embeddings.shape)
```

# 5. Store the embeddings into Faiss

```python
import faiss
import numpy as np

embeddings = np.array(embeddings, dtype="float32")

index = faiss.IndexFlatL2(embeddings.shape[1])

index.add(embeddings)

print("Total Vecore Stores : ",index.ntotal)
```

# 6. Retrive Relevent Information

```python
question = "What are the common warning signs of machine problems?"

query_embedding = model.encode([question])

query_embedding = np.array(query_embedding, dtype="float32")

distances, indices = index.search(query_embedding, k=3)

for i in indices[0] :
    print("Source : ",chunks[i]["source"])
    print(chunks[i]["text"])
    print("-"*50)
```

# 8. Test model with multiple questions

```python
def ask_question(question):
    query_embedding = model.encode([question])

    query_embedding = np.array(query_embedding, dtype="float32")

    distances, indices = index.search(query_embedding, k=3)

    context = ""

    for i in indices[0] :
        context += (
            f"\nSource :{chunks[i]['source']}\n"
            f"{chunks[i]["text"]}\n"
        )

    prompt = f"""
    You are a Manufacturing AI Assistant.

    Answer the question using only the provided context.
    If the answer is not available in the context, say:
    "The provided documents do not contain enough information."

    context : {context}
    Question : {question}
    Answer :
    """
    response = client.models.generate_content(
        model = "gemini-flash-latest",
        contents = prompt
    )

    return response.text
```

```python
question = "What are the common warning signs of machine problems?"

answer = ask_question(question)

print(answer)
```

```python
questions = [
    "What are the steps for machine setup?",
    "What safety precautions should workers follow?",
]

for i in questions :
    print("\nQuestion : ",i)
    print("Answer :\n",ask_question(i))
    print("-"*50)
```

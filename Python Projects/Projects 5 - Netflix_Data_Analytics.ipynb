```python
# Content rating distribustion
colors = plt.cm.tab20.colors
df["rating"].value_counts().sort_index().plot(kind="bar",color=colors)
plt.title("Content rating distribustion")
plt.xlabel("Content")
plt.ylabel("Count")
plt.show()
```

```python
# Movie Duration Distribustions 
movie_df = df[df["type"] == "Movie"].copy()
movie_df["duration_int"] = movie_df["duration"].str.extract(r"(\d+)").astype(float)
plt.figure(figsize=(10,5))
sns.histplot(movie_df["duration_int"],bins=30,kde=True,color="purple")
plt.title("Movie Duration Distribustion")
plt.xlabel("Duration in minutes")
plt.ylabel("Frequency")
plt.show()
```

```python
# Top 10 most common genres on netflix
from collections import Counter

all_genres = ",".join(df["listed_in"].dropna()).split(",")
genre_count = Counter([genre.strip() for genre in all_genres])
top_genres = genre_count.most_common(10)
label,value = zip(*top_genres)
plt.figure(figsize=(20,6))

colors = plt.cm.tab20.colors

sns.barplot(x=list(label),y=list(value),hue=list(label),legend=True)

plt.title('Top 10 Most Common Genres')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```

```python
# Top director on Netflix
top_directors = df[df["director"] != "Unknown"]["director"].value_counts().head(10)
plt.figure(figsize=(15,5))
sns.barplot(x=top_directors.values,y=top_directors.index,palette="viridis")
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Directors")
plt.show()
```

```python
# Top director on Netflix
top_directors = df[df["director"] != "Unknown"]["director"].value_counts().head(10)
plt.figure(figsize=(15,5))
sns.barplot(x=top_directors.values,y=top_directors.index,hue=top_directors.index,legend=False)
plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Directors")
plt.show()
```

# Key findings
1.Movies are more widely represented than TV Shows in the Netflix dataset.<br>
2.The number of Netflix productions increased sharply in 2021.<br>
3.Netflix has the most content from the United States, followed by India in second place and the Japan in third place.<br>
4.In the overall rating distribution bar chart, we can see that most of the shows have TV-MA ratings, indicating that the majority of the shows are for mature audiences, with TV-14 ratings coming in second.<br>
5.Most movies on Netflix have a runtime of around 90–120 minutes.<br>
6.As shown in the bar graph, Dramas are the most common genre followed by International Movies on second place and Comedies hold the third place.<br>
7.Toshiya Shinohara is the top director on Netflix, with 4 titles, followed by Masahiko Murata in second place with 3 titles.

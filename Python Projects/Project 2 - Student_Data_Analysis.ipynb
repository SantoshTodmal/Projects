<b> 1. Student Marks Analysis

```python
students = {
    "Amit": 85,
    "Neha": 92,
    "Rahul": 76,
    "Priya": 88,
    "Rohan": 67,
    "Sneha": 95
}
```

```python
# 1.Display all students.
print(students)
```

```python
# Find highest and lowest mark
highest = 0

for name in students:
    if students[name] > highest :
        highest = students[name]
print(name,highest)
```

```python
lowest = 100
for name in students:
    if students[name] < lowest :
        lowest = students[name]
print(name,lowest)
```

```python
# Calculate average marks
total = 0
for values in students.values():
    total += values
avg = total/len(students.values())
print(f'{avg:.2f}')
```

```python
# Count students above 80
students = {
    "Amit": 85,
    "Neha": 92,
    "Rahul": 76,
    "Priya": 88,
    "Rohan": 67,
    "Sneha": 95
}
count = 0
for values in students.values():
    if values > 80 :
        count +=1
print(count)
```

```python
# student names and their marks if they scored above 80.
for name in students:
    if students[name] > 80:
        print(name,students[name])
```

```python
# Find the Highest Marks and Student Name
students = {
    "Amit": 85,
    "Neha": 92,
    "Rahul": 76,
    "Priya": 88,
    "Rohan": 95,
    "Sneha": 67
}
highest = 0
highest_students = " "
for name in students:
    if students[name] > highest:
        highest = students[name]
        highest_students = name
print(f'Highest marks : {highest} \nStudent : {highest_students}')
```

```python
# lowest marks and student name.
lowest = 100
lowest_student = " "
for name in students:
    if students[name] < lowest:
        lowest = students[name]
        lowest_student  = name
print(f'lowest : {lowest} \nStudent : {lowest_student}')
```

```python
# Print the grade of every student.
for name in students:
    marks = students[name]
    if marks >= 90:
        print(f'{name} : {marks} Grade A')
    elif marks >= 80:
        print(f'{name} : {marks} Grade B')
    elif marks >= 70:
        print(f'{name} : {marks} Grade C')
    elif marks >= 60:
        print(f'{name} : {marks} Grade D') 
    else:
        print(f'{name} : {marks} Grade F') 
```

```python
# Count Students in Each Grade
grade_a = 0
grade_b = 0
grade_c = 0
grade_d = 0
grade_f = 0
for name in students:
    marks = students[name]
    
    if marks >= 90 :
        grade_a += 1
    elif marks >= 80:
        grade_b +=1
    elif marks >= 70:
        grade_c +=1
    elif marks >= 60:
        grade_d +=1
    elif marks < 60:
        grade_f +=1
        
print(f'Grade A :{grade_a}')
print(f'Grade B :{grade_b}')
print(f'Grade C :{grade_c}')
print(f'Grade D :{grade_d}')
print(f'Grade F :{grade_f}')
```

```python
students = {
    "Amit": 85,
    "Neha": 92,
    "Rahul": 76,
    "Priya": 88,
    "Rohan": 95,
    "Sneha": 67
}

count = len(students)
total = 0
stu_count = 0
highest_marks = 0
lowest_marks = 100
highest_student = " "
lowest_student = " "

for name in students:
    if students[name] > highest_marks:
        highest_marks = students[name]
        highest_student = name
    if students[name] < lowest_marks:
        lowest_marks = students[name]
        lowest_student = name
    if students[name] > 80:
        stu_count +=1
for values in students.values():
    total += values
avg = total/count

print("========== STUDENT REPORT ==========")
print("Total Students : ", count)
print(f'Highest Marks : {highest_marks} ({highest_student})')
print(f'Lowest Marks : {lowest_marks} ({lowest_student})')   
print(f"Average Marks : {avg:.2f}")
print(f"Students Above 80 : {stu_count}")
print("===================================")
```

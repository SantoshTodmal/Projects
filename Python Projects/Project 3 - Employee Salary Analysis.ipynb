<b>Employee Salary Analysis

```python
employee = ["Amit", "Priya", "Rahul", "Neha", "Karan"]
salary = [35000, 52000, 48000, 61000, 45000]
```

```python
# Find the total salary paid to all employees.
total = 0
for i in salary:
    total += i
print("Total salary paid to all employees are :",total)
```

```python
# Find the average salary.
total = 0
for i in salary:
    total += i
avg = total/len(salary)
print("Average salary :",avg)
```

```python
# Find the highest salary.
highest = salary[0]
for i in salary:
    if i > highest:
        highest = i
print("Highest salary paid :",highest)
```

```python
# Find the lowest salary.
lowest = salary[0]
for i in salary:
    if i < lowest:
       lowest = i
print("Lowest salary paid :",lowest)
```

```python
# Display the employee who has the highest salary.
employee = ["Amit", "Priya", "Rahul", "Neha", "Karan"]
salary = [35000, 52000, 48000, 61000, 45000]
highest_salary = salary[0]
highest_employee = employee[0]
for i in range(len(salary)):
    if salary[i] > highest_salary:
        highest_salary = salary[i]
        highest_employee  = employee[i]
    
print(highest_employee ,":",highest_salary)
```

```python
# Display the employee who has the lowest salary.
lowest_salary = salary[0]
lowest_employee = employee[0]
for i in range(len(salary)):
    if salary[i] < lowest_salary:
        lowest_salary = salary[i]
        lowest_employee  = employee[i]
    
print(lowest_employee ,":",lowest_salary)
```

```python
# Count how many employees earn more than ₹50,000.
count = 0
for i in salary:
    if i > 50000:
        count += 1
print("Employees earn more than ₹50,000 :",count)
```

```python
# Display all employees earning less than ₹50,000.
for i in range(len(salary)):
    if salary[i] < 50000:
        print(employee[i] ,":", salary[i])
```

```python
# Give a salary category.
for i in range(len(salary)):
    if salary[i] >= 60000:
        print(employee[i],":","High")
    elif salary[i] >= 45000:
        print(employee[i],":","Medium")
    else:
        print(employee[i],":","Low")
```

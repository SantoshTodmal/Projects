```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer", "Speaker"]

price = [55000, 800, 1500, 12000, 9000, 2500]

quantity = [8, 35, 20, 12, 6, 15]
```

```python
# Calculate the total revenue.
total_revenue = 0
for i in range(len(products)):
    revenue = price[i]*quantity[i]
    total_revenue += revenue
print("Total revenue :",total_revenue)
```

```python
# Display the revenue of each product.
for i in range(len(products)):
    revenue = price[i]*quantity[i]
    
    print(products[i],":",revenue)
```

```python
# Find the product with the highest revenue.
highest_revenue = price[0] * quantity[0]
highest_product = products[0]
for i in range(len(products)):
    revenue  = price[i]*quantity[i]
    
    if revenue > highest_revenue:
        highest_revenue = revenue
        highest_product = products[i]
print("Highest Revenue Product")
print(highest_product,":",highest_revenue)
```

```python
# Find the product with the lowest revenue.
lowest_revenue = price[0] * quantity[0]
lowest_product = products[0]
for i in range(len(products)):
    revenue  = price[i]*quantity[i]
    
    if revenue < lowest_revenue:
        lowest_revenue = revenue
        lowest_product = products[i]
print("Lowest Revenue Product")
print(lowest_product,":",lowest_revenue)
```

```python
# Count how many products generated revenue greater than ₹50,000.
count = 0
for i in range(len(products)):
    revenue  = price[i]*quantity[i]
    if revenue > 50000:
        count +=1 
print("products generated revenue greater than ₹50,000:",count)
```

```python
# Display all products whose quantity sold is greater than 10.
for i in range(len(products)):
    if quantity[i] > 10:
        print(products[i])
```

```python
# Display products whose price is greater than ₹10,000.
for i in range(len(products)):
    if price[i] > 10000:
        print(products[i])
```

```python
# Give each product a category based on its price.

for i in range(len(products)):
    if price[i] >= 20000:
        print(products[i],":","Premium")
    elif price[i] >= 5000:
        print(products[i],":","Standard")
    else:
        print(products[i],":","Budget")
```

```python
# Find the average revenue per product.
total_revenue = 0
for i in range(len(products)):
    revenue = price[i]*quantity[i]
    total_revenue += revenue
average_revenue = total_revenue/len(products)
print("Average Revenue :",average_revenue)
```

```python
print("Product Sales Report")
print("--------------------")
count = 0
total_revenue = 0
highest_revenue = price[0] * quantity[0]
highest_product = products[0]
lowest_revenue = price[0] * quantity[0]
lowest_product = products[0]

for i in range(len(products)):
    revenue = price[i]*quantity[i]
    total_revenue += revenue
    print(products[i],":",revenue)
    
    if revenue > highest_revenue:
        highest_revenue = revenue
        highest_product = products[i]
    if revenue < lowest_revenue:
        lowest_revenue = revenue
        lowest_product = products[i]

    if revenue > 50000:
        count +=1 
        
avg_revenue = total_revenue/len(products)
print("\nTotal revenue :",total_revenue)
print("Average Revenue :",avg_revenue)
print("Highest Revenue Product :", highest_product,":",highest_revenue)
print("Lowest Revenue Product :", lowest_product,":",lowest_revenue)
print("Products earning above ₹50000 :",count)
    
```

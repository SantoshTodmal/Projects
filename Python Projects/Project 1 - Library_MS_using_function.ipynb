```python
def add_book():
    
    book = {
        "Book ID": int(input("Enter book ID")),
        "Book Title":  input("Book Title"),
        "Author Name": input("Author Name"),
        "Status": "Available",
        "Student Name":""}
    books.append(book)
    
    print("Book added to the library successfully")
```

```python
def display_books():
    if len(books) == 0:
        print("No books available")
        return
    
    for book in books:
        print("Book ID :",book["Book ID"])
        print("Book Title :",book["Book Title"])
        print("Author Name:",book["Author Name"])
        print("Status:",book["Status"])
        print("Student Name:",book["Student Name"])
        print("="*40)
```

```python
def search_book():
    while True:
        print("1.Search book by Book ID")
        print("2.Search book by Book Title")
        print("3.Search book by Author Name")
        print("4.Back to main menu")
        print("="*40)
        search_choice = int(input("Enter your book search choice"))
            
        if search_choice == 1:
            Book_found = False
            book_id = int(input("Enter Book ID :"))
                
            for book in books:
                if book_id == book["Book ID"] :
                        print("Book ID :",book["Book ID"])
                        print("Book Title :",book["Book Title"])
                        print("Author Name:",book["Author Name"])
                        print("Status:",book["Status"])
                        print("Student Name:",book["Student Name"])
                        print("="*40)
                        Book_found = True
                        print("="*40)
            if Book_found == False:
                print("Book not found")

        elif search_choice == 2:
            Book_found = False
            book_title = input("Enter a Book Title")
            for book in books:
                if book_title.lower() in book["Book Title"].lower():
                        print("Book ID :",book["Book ID"])
                        print("Book Title :",book["Book Title"])
                        print("Author Name:",book["Author Name"])
                        print("Status:",book["Status"])
                        print("Student Name:",book["Student Name"])
                        print("="*40)
                        Book_found = True
            if Book_found == False:
                print("Book not found")
                print("="*40)
                    
        elif search_choice == 3:
            Book_found = False
            author_name = input("Enter a Author Name")
            for book in books:
                if author_name.lower() in book["Author Name"].lower():
                        print("Book ID :",book["Book ID"])
                        print("Book Title :",book["Book Title"])
                        print("Author Name:",book["Author Name"])
                        print("Status:",book["Status"])
                        print("Student Name:",book["Student Name"])
                        print("="*40)
                        Book_found = True
            if Book_found == False:
                print("Book not found")
                print("="*40)
                    
        elif search_choice == 4:
            print("Return to the main menu")
            break
            
        else:
            print("Invalid choice")
```

```python
def update_book():
    
    book_id = int(input("Enter a Book ID to update :"))
    Book_found = False
    
    for book in books:
        if book_id == book["Book ID"]:
            print("Book Found")
            print(book)
            print("="*40)
                
            new_title = input("Enter a new Book Title :")
            new_author = input("Enter a new Author Name :")

            book["Book Title"] = new_title
                
            book["Author Name"] = new_author
            print("Book details update successfully")
            print(book)
                                   
            Book_found = True
            break
                
    if Book_found == False:
        print("Book not found")
        print("="*40)
```

```python
def delete_book():
    
    book_id = int(input("Enter a Book ID to delete :"))
    Book_found = False
    
    for book in books:
        if book_id == book["Book ID"]:
            books.remove(book)
            Book_found = True
            print("Book deleted successfully")
            print(book)
            print("="*40)
            break
                
    if Book_found == False:
        print("Book not found")
        print("="*40)
```

```python
def issue_book():
    
    book_id = int(input("Enter book id to issued :"))
    book_found = False
    
    for book in books:
        if book["Book ID"] == book_id:
            book_found = True

            if book["Status"] == "Available":
                book["Status"] = "Issued"
                book["Student Name"] = input("Enter a student name :")
                print("Book issued successfully")
            else:
                print("Book is already issued")

            break
                    
    if book_found == False:
        print("Book not found")
    print("="*40)
```

```python
def return_book():
    book_id = int(input("Enter book id of issued book to return:"))
    book_found = False

    for book in books:
        if book["Book ID"] == book_id:
            book_found = True
            if book["Status"] == "Issued":
                book["Status"] = "Available"
                book["Student Name"] = ""
                print("Book returned successfully")
            else:
                print("Book is already available")
            break
            
    if book_found == False:
        print("Book not found")
    print("="*40)
```

```python
def display_availble_books():
    
    available_found = False
    for book in books:
        if book["Status"] == "Available":
            print("Book ID :",book["Book ID"])
            print("Book Title :",book["Book Title"])
            print("Author Name:",book["Author Name"])
            print("Status:",book["Status"])
            print("Student Name:",book["Student Name"])
            print("="*45)
            available_found = True
                
    if available_found == False:
        print("No books are available")
    
    print("="*40)
```

```python
def display_issued_books():
    
    issued_found = False
    for book in books:
        if book["Status"] == "Issued":
            print("Book ID :",book["Book ID"])
            print("Book Title :",book["Book Title"])
            print("Author Name:",book["Author Name"])
            print("Status:",book["Status"])
            print("Student Name:",book["Student Name"])
            print("="*45)
            issued_found = True
            
    if issued_found == False:
        print("No books are issued")
    
    print("="*40)
```

```python
def main_menu():
    
    print("Library Management System")
    print("="*40)
    
    while True:
        print("1.Add a book")
        print("2.Display all books")
        print("3.Search book")
        print("4.Update book details")
        print("5.Delete book from library")
        print("6.Issue a book")
        print("7.Return a book")
        print("8.Display available books")
        print("9.Display issued books")
        print("10.Exit the application")
        print("="*40)   
        
        choice = int(input("Enter your choice :"))
    
        if choice == 1:
            add_book()
    
        elif choice == 2:
            display_books()
    
        elif choice == 3:
            search_book()
    
        elif choice == 4:
            update_book()
      
        elif choice == 5:
            delete_book()
        
        elif choice == 6:
            issue_book()
        
        elif choice == 7:
            return_book()
    
        elif choice == 8:
            display_availble_books()
    
        elif choice == 9:
            display_issued_books()
    
        elif choice == 10:
            print("You exited the application safely.\nThank You!")
            break
        else:
            print("Invalid choice. Please try again.")
```
